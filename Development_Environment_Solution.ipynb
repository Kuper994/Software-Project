You are running a zoo, and you need to maintain its file system using the Linux command line.

## 1. Set up the zoo

Create the following directory structure:

```text
zoo/
├── animals/
├── staff/
└── reports/
```

Navigate into `zoo/` and verify your current location.

````python
%%bash
mkdir -p zoo/animals zoo/staff zoo/reports
````

## 2. Add the animals

Inside `animals/`, create three files:

```text
lions.txt
penguins.txt
elephants.txt
```

Each file should contain the names of the animals in that enclosure.

For example, `lions.txt` should contain:

```text
Simba
Nala
Mufasa
```

Create similar lists for the penguins and elephants.


````python
%cd zoo
````

````python
%%bash
echo 'Simba' > animals/lions.txt
echo 'Nala' >> animals/lions.txt
echo 'Mufasa' >> animals/lions.txt

echo 'Simba' > animals/penguins.txt
echo 'Nala' >> animals/penguins.txt
echo 'Mufasa' >> animals/penguins.txt

echo 'Simba' > animals/elephants.txt
echo 'Nala' >> animals/elephants.txt
echo 'Mufasa' >> animals/elephants.txt
````

## 6. Create the daily report

Create:

```text
reports/daily_animal_report.txt
```

The report should contain:

```text
Zoo Inventory Summary
Lions: <number>
Penguins: <number>
Elephants: <number>
Total animals: <number>
```

Use shell commands to create the file and add its contents. Then display the report.


````python
%%bash
echo "Zoo Inventory Summary"
{
  echo "Lions: $(cat animals/lions.txt | wc -l)"
  echo "Pinguins: $(cat animals/penguins.txt | wc -l)"
  echo "Elephants: $(cat animals/elephants.txt | wc -l)"
  echo "Total animals: $(cat animals/*.txt | wc -l)"
} > reports/daily_animal_report.txt

cat reports/daily_animal_report.txt
````

## 11. Run a zoo report

Create:

```text
reports/zoo_report.sh
```

The script should:

1. Print `Zoo Report`.
2. Display the contents of all animal files.
3. Display the contents of the daily report.

Make the script executable and run it.

**Question:** Why do we use `./zoo_report.sh` to run the script?


````python
# first line is needed in order to run it as bash script
content = """#!/bin/bash

echo "Zoo Report"

for file in animals/*; do
    if [ -f "$file" ]; then
        echo "$file"
        cat "$file"
    fi
done

echo "Daily Report"
cat reports/daily_animal_report.txt
"""

with open("reports/zoo_report.sh", "w") as f:
    f.write(content)


!chmod +x reports/zoo_report.sh
!./reports/zoo_report.sh
````
